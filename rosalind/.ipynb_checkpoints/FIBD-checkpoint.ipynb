```python
# n is number of iterations
# k is number of offspring pairs a rabbit pair produces
# m is number of months that rabbits die after

def get_result(n,k,m):
    iterations = [1,1]
    total = 1
    count = 1
    for i in range(n - 2):
        total = iterations[count] + (iterations[count - 1] * k)
        if count > m - 1:
            total = total - iterations[count - m - 1]
        iterations.append(total)
        count = count + 1
    print(iterations)
    return iterations[-1]
```

```python
get_result(6,1,3)
```
