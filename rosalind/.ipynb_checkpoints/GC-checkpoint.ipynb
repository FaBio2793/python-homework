```python
def gc(dna):
    gc_content = ((dna.count("G") + dna.count("C")) / len(dna))*100
    return gc_content
```

```python
gc("GCGCGGGAAATATTA")
```

```python
sequences = {}

with open("./rosalind_gc(10).txt", "r") as rosalind:
    for line in rosalind:
        if line[0] == ">":
            last_id = line[1:].rstrip()
            sequences[last_id] = ""
        else:
            sequences[last_id] += line.rstrip()
```

```python
print(sequences)
```

```python
gc_contents = {}
for seq_id, seq in sequences.items():
    gc_contents[seq_id] = gc(seq)
```

```python
print(gc_contents)
```

```python
max_gc = max(gc_contents.values())

for seq_id in gc_contents:
    if gc_contents[seq_id] == max_gc:
        print(seq_id)
        print(round(max_gc, 6))
```
