```python
def get_result(n,k):
    iterations = [1,1]
    total = 1
    count = 1
    for i in range(n - 2):
        total = iterations[count] + (iterations[count - 1] * k)
        iterations.append(total)
        count = count + 1
    print(iterations)
    print(iterations[-1])
```

```python
get_result(5,3)
```
