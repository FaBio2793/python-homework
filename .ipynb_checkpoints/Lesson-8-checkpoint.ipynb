```python
for letter in [0,1,2]:
    print(str(words[letter])[0])
```

```python
acronym = ""
```

```python
for word in words:
    acronym = acronym + word[0]
print(acronym)
```

```python
for word in words:
    acronym = acronym + word[0].upper()
print(acronym)
```

```python
cumulative = []
total = 0
data = [1,2,2,5]
for number in data:
    total = total + number
    cumulative.append(total)
print(cumulative)
```

```python
message = ""
for number in range(10):
    # use a if the number is a multiple of 3, otherwise use b
    if (number % 3) == 0:
        message = message + "a"
    else:
        message = message + "b"
print(message)
```
