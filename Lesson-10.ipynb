```python
# revised version
import random
for i in range(10):

    # simulating the mass of a chicken egg
    # the (random) mass will be 70 +/- 20 grams
    mass = 70 + 40.0 * (2.0 * random.random() - 1.0)

    print(mass, get_egg_label(mass))
```

```python
# egg sizing machinery prints a label
def get_egg_label(mass):
    egg_label = "unlabelled"
    if mass >= 85 and mass < 90:
        egg_label = "jumbo"
    elif mass >= 70 and mass < 85:
        egg_label = "large"
    elif mass < 70 and mass >= 55:
        egg_label = "medium"
    elif mass > 90:
        egg_label = "very heavy, maybe dirty?"
    elif mass < 50:
        egg_label = "too light, probably spoiled"
    elif mass < 55 and mass > 50:
        egg_label = "small"
    return egg_label
```

```python
def logistic_map(x, r):
    return r * x * (1-x)
```

```python
r = 1.0
```

```python
population = [initial_population]
```

```python
for t in range(t_final):
    population.append(logistic_map(population[t], r))
```

```python
def iterate(initial_population, t_final, r):
    develop = [initial_population]
    for t in range(t_final):
        develop.append(logistic_map(develop[t], r))
    return develop
```

```python
iterate(0.7, 20, 1)
```
